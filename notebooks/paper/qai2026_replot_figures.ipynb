# QAI 2026 Figure Replot Notebook

This notebook regenerates paper figures from saved artifacts. It does not train models.

```python
from pathlib import Path
import subprocess
import sys

REPO_ROOT = Path.cwd().parents[1] if Path.cwd().name == 'paper' else Path.cwd()
print(REPO_ROOT)
```

## Adjustable Style Controls

```python
FONT_SIZE = 10
FORMATS = ['pdf', 'png']
```

## Seed Current Curated Artifacts

Run this when the curated JSON artifact is missing or after updating notebook-derived scalar values.

```python
subprocess.run(
    [sys.executable, str(REPO_ROOT / 'scripts' / 'qai2026_seed_artifacts.py')],
    cwd=REPO_ROOT,
    check=True,
)
```

## Replot Adjustable Figures

```python
subprocess.run(
    [
        sys.executable,
        str(REPO_ROOT / 'scripts' / 'qai2026_replot_figures.py'),
        '--font-size', str(FONT_SIZE),
        '--formats', *FORMATS,
    ],
    cwd=REPO_ROOT,
    check=True,
)
```

## Extract Notebook Preview Figures

These previews come from embedded notebook output cells. Use them for layout checks only; final figures should come from the replot script when possible.

```python
subprocess.run(
    [sys.executable, str(REPO_ROOT / 'scripts' / 'qai2026_extract_notebook_images.py')],
    cwd=REPO_ROOT,
    check=True,
)
```

## Write LaTeX Tables

```python
subprocess.run(
    [sys.executable, str(REPO_ROOT / 'scripts' / 'qai2026_write_tables.py')],
    cwd=REPO_ROOT,
    check=True,
)
```
